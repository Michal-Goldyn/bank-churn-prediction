```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
```

```python
df = pd.read_csv("../../data/Customer-Churn-Records.csv")
```

```python
pd.crosstab(df["IsActiveMember"], df["Exited"])
```

```python
contingency_table = pd.crosstab(df["IsActiveMember"], df["Exited"])
chi2, p, dof, expected = stats.chi2_contingency(contingency_table)
print(f"p-value: {p}")
```

```python
cramers_v = stats.contingency.association(contingency_table, method="cramer")
print(f"Cramér's V: {cramers_v}")
```

## Hypothesis 1: Active members churn less frequently

Test used: _Chi-square test_

The test showed a p-value close to 0, meaning the result is statistically significant 
- the relationship between activity and churn is not random. However, with 10,000 
customers, almost everything will be statistically significant. Therefore, Cramér's V 
was also calculated to measure the strength of the relationship.

Cramér's V = 0.156 - the effect is weak.

**Conclusion:** Active members do churn less frequently, but the relationship 
is relatively weak. Being inactive alone is not a strong predictor of churn.

```python
exited = df[df["Exited"] == 1]["CreditScore"]
stayed = df[df["Exited"] == 0]["CreditScore"]
```

```python
exited
```

```python
stayed
```

```python
t_stat, p_value = stats.ttest_ind(exited, stayed)
print(f"p-value: {p_value}")
print(f"Mean CreditScore exited: {exited.mean():.2f}")
print(f"Mean CreditScore stayed: {stayed.mean():.2f}")
```

```python
n1, n2 = len(exited), len(stayed)
std1, std2 = exited.std(), stayed.std()
pooled_std = np.sqrt(((n1-1)*std1**2 + (n2-1)*std2**2) / (n1+n2-2))
cohens_d = (exited.mean() - stayed.mean()) / pooled_std
print(f"Cohen's d: {cohens_d}")
```

## Hypothesis 2: Higher CreditScore reduces churn probability

Test used: _t-test (independent samples)_

Mean CreditScore - churned customers: 645.41, retained customers: 651.84.

The p-value (0.0074) is below 0.05, meaning the difference is statistically significant. 
However, the effect size (Cohen's d = 0.066) is negligible - far below the threshold 
for even a small effect.

Cohen's d reference: 0.2 = small effect, 0.5 = medium effect, 0.8+ = large effect.

**Conclusion:** Despite statistical significance, CreditScore has virtually no practical 
impact on churn. With a 6-point difference between groups, this finding has limited 
business value - the bank should not prioritize credit score as a churn indicator.

## Hypothesis 3: Gender affects churn

Test used: _Chi-square test_

Female customers churn more often (25% vs 16% for males, as seen earlier in the EDA). 
The p-value (2.93e-26) confirms this is statistically significant.

However, Cramér's V = 0.106 - even weaker than the effect found for IsActiveMember (0.156).

**Conclusion:** Gender has a statistically significant but weak relationship with churn. 
On its own, gender is not a strong predictor of customer churn.

## Hypothesis 4: Age differs significantly between churned and retained customers

Test used: _t-test (independent samples)_

Mean age - churned customers: 44.84 years, retained customers: 37.41 years.

The p-value is essentially 0, confirming statistical significance. More importantly, 
Cohen's d = 0.739 - a large effect size, close to the 0.8 threshold.

**Conclusion:** Customers who leave the bank are on average 7 years older than those 
who stay. This is the strongest finding among all tested hypotheses - age is a 
meaningful and practically significant factor in churn.

## Hypothesis 5: Customers with higher Balance churn more often

Test used: _t-test (independent samples)_

Mean Balance - churned customers: 91,109, retained customers: 72,743.

The p-value is essentially 0, confirming statistical significance. Cohen's d = 0.296, a small-to-medium effect, noticeably weaker than age (0.739) but stronger than CreditScore (0.066) or gender (0.106).

**Conclusion:** Customers who leave tend to have higher balances than those who stay, a moderate but meaningful difference. This may relate to higher-balance customers 
having more bargaining power or being more attractive targets for competing banks.
