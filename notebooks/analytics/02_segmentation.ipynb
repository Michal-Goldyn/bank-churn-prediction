```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../../data/Customer-Churn-Records.csv")
```

```python
geography_mean = df.groupby(["Geography"])["Exited"].mean()
geography_mean
```

```python
sns.barplot(x=geography_mean.index, y=geography_mean.values)
plt.title("Churn Rate by Geography")
plt.xlabel("Geography")
plt.ylabel("Churn Rate")
plt.savefig("../../reports/figures/Churn_Rate_by_Geography.png", dpi=150, bbox_inches="tight")
plt.show()
```

Germany has twice the churn rate compared to France and Spain (32% vs 16%). 
The bank should review its offering in Germany - local conditions or product terms 
may not be attractive enough for German customers.

```python
age_group = df["AgeGroup"] = pd.cut(df["Age"], bins=[18, 30, 45, 60, 100], labels=["18-30", "30-45", "45-60", "60+"])
age_group.value_counts()
```

```python
age_churn = df.groupby(["AgeGroup"])["Exited"].mean()
age_churn
```

```python
sns.barplot(x=age_churn.index, y=age_churn.values)
plt.title("Churn Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Churn Rate")
plt.savefig("../../reports/figures/Churn_Rate_by_AgeGroup.png", dpi=150, bbox_inches="tight")
plt.show()
```

The 45-60 age group has the highest churn rate at 51%, which is very high. 
This means that 1 in 2 customers in this age group leaves the bank.

```python
age_x_geography_churn = df.groupby(["AgeGroup", "Geography"])["Exited"].mean().unstack()
age_x_geography_churn
```

```python
sns.heatmap(age_x_geography_churn, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Churn Rate by Age Group and Geography")
plt.xlabel("Geography", labelpad=15, color="blue")
plt.ylabel("Age Group", labelpad=15, color="blue")
plt.savefig("../../reports/figures/Churn_Rate_by_Age_and_Geography.png", dpi=150, bbox_inches="tight")
plt.show()
```

The heatmap clearly shows a red segment - the 45-60 age group in Germany 
has the highest churn rate at 67%. This is the most critical customer segment 
the bank should focus on.

```python
products_churn = df.groupby("NumOfProducts")["Exited"].mean()
products_churn
```

```python
sns.barplot(x=products_churn.index, y=products_churn.values)
plt.title("Churn Rate by Number of Products")
plt.xlabel("Number of Products")
plt.ylabel("Churn Rate")
plt.savefig("../../reports/figures/Churn_Rate_by_NumOfProducts.png", dpi=150, bbox_inches="tight")
plt.show()
```

Customers with 2 products are the most loyal (8% churn), while those with 3+ products 
churn at an alarming rate — 83% for 3 products and 100% for 4 products.

Possible explanations:
* The more products a customer has, the more they pay for additional services - 
  costs may become too high
* Products 3 and 4 may have worse conditions (though we cannot be certain 
  without knowing exactly what these products are)

The bank should investigate what products 3 and 4 are and why they drive customers away.

## Recommendations for the bank director

If I were the bank director, I would immediately investigate:

1. **German market, especially the 45-60 age group** - with a 67% churn rate, 
   this is the most critical segment. Are product terms competitive enough 
   compared to local German banks?

2. **Customers with 3+ products** - 83-100% churn rate is alarming. 
   Are there unreasonable requirements or excessive fees for these products? 
   This needs urgent review.
